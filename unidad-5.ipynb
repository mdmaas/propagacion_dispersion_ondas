# Unidad 5: El método adjunto

En la unidad anterior vimos cómo **ajustar un modelo de velocidades** contra observaciones
cuando el modelo estaba parametrizado por un **polinomio de unos pocos coeficientes**. En
ese régimen, armar la matriz de sensibilidades (las derivadas del estado respecto de cada
parámetro) con diferencias finitas y resolver el problema de mínimos cuadrados con
**Gauss–Newton** es perfectamente viable: si hay $p$ parámetros, hacemos del orden de $p$
evaluaciones del modelo directo.

El objetivo de esta unidad es entender qué hacer cuando el número de parámetros es enorme:

* en **full-waveform inversion (FWI)** el modelo de velocidades vive en una malla y puede
  tener **millones** de parámetros;
* en **optimización de formas** los grados de libertad son los nodos de la frontera;
* en **diseño inverso** queremos encontrar una permitividad $\varepsilon(\mathbf x)$ en toda
  una región.

En estos casos calcular $\dfrac{dx}{dm}$ columna por columna es prohibitivo. La herramienta
que resuelve el problema es el **método adjunto**, que calcula el gradiente completo con un
costo **independiente del número de parámetros**.

## Contenido

1. **El método adjunto en dimensión finita** — la base: multiplicadores de Lagrange.
2. **Un ejemplo de control óptimo discreto.**
3. **La ecuación de Poisson:** *discretizar primero*, aplicar el adjunto y reconocer la
   ecuación adjunta.
4. **Focalización con la ecuación de Helmholtz.**
5. **El punto de vista en dimensión infinita:** Lagrange en espacios de Hilbert, derivadas de
   Gateaux y la comparación entre *optimizar-y-discretizar* y *discretizar-y-optimizar*.
6. **Comentarios finales y ejercicios.**

Seguiremos siempre el mismo patrón: escribimos el problema directo, construimos el
Lagrangiano, derivamos la **ecuación adjunta** y obtenemos el **gradiente**. En los problemas
con EDP **discretizamos primero** y solo después interpretamos el sistema adjunto discreto
como la discretización de una ecuación diferencial adjunta.

> **Nota sobre la numeración.** Las ecuaciones se numeran automáticamente y se pueden
> referenciar con hipervínculos; cada referencia de la forma `{eq}`etiqueta`` salta a la
> ecuación correspondiente.
