# 1. El método adjunto en dimensión finita

## 1.1 El problema

Consideremos un problema de optimización con **restricciones de igualdad**

$$
\min_{x,m}\; J(x,m)
\qquad\text{sujeto a}\qquad
g(x,m)=0,
\qquad x\in\mathbb R^n,\quad m\in\mathbb R^p,
$$

donde $g:\mathbb R^n\times\mathbb R^p\to\mathbb R^n$. Interpretamos

* $x$ como la **variable de estado** (la solución de un problema directo), y
* $m$ como los **parámetros de diseño o control** (lo que queremos optimizar).

En las aplicaciones de interés $n$ y $p$ son grandes; típicamente $p\sim10^6$–$10^9$.

## 1.2 Reducción a un problema sin restricciones

Supongamos que para cada $m$ existe una única solución $x(m)$ de $g(x(m),m)=0$ y que
$D_xg$ (la matriz jacobiana de $g$ respecto de $x$) es invertible. Por el **teorema de la
función implícita**, $x(m)$ es diferenciable y el problema es equivalente al problema **sin
restricciones**

$$
\min_m\; \hat J(m) = J\big(x(m),m\big).
$$

Queremos minimizarlo con un método de descenso, lo que requiere el gradiente
$\dfrac{d\hat J}{dm}\in\mathbb R^{1\times p}$. Por la regla de la cadena,

$$
\frac{d\hat J}{dm}
=\underbrace{\frac{\partial J}{\partial m}}_{1\times p}
+\underbrace{\frac{\partial J}{\partial x}}_{1\times n}\,
 \underbrace{\frac{dx}{dm}}_{n\times p}.
$$ (eq:chain)

Las derivadas parciales de $J$ son sencillas de calcular. La dificultad está en la **matriz
de sensibilidad** $\dfrac{dx}{dm}$, que proviene de la restricción. En efecto, derivando
$g(x(m),m)=0$ respecto de $m$ y usando la regla de la cadena,

$$
D_xg\,\frac{dx}{dm} + D_mg = 0
\qquad\Longrightarrow\qquad
\frac{dx}{dm} = -\big(D_xg\big)^{-1}D_mg .
$$ (eq:sensitivity)

Sustituyendo {eq}`eq:sensitivity` en {eq}`eq:chain`,

$$
\frac{d\hat J}{dm}
= \frac{\partial J}{\partial m}
- \frac{\partial J}{\partial x}\big(D_xg\big)^{-1}D_mg .
$$

El problema de esta última expresión es el producto $\big(D_xg\big)^{-1}D_mg$: equivale a
resolver **un sistema lineal por cada columna de $D_mg$**, es decir, $p$ resoluciones. El
costo es $O(p)$, inaceptable cuando $p$ es enorme.

## 1.3 El Lagrangiano y la ecuación adjunta

Introducimos un multiplicador de Lagrange $\lambda\in\mathbb R^n$ y formamos el
**Lagrangiano**

$$
\mathcal L(x,m,\lambda)=J(x,m)+\lambda^{T}g(x,m).
$$ (eq:lagrangian)

En un óptimo buscamos un **punto estacionario** de $\mathcal L$. Las condiciones son:

**(a) Estacionariedad respecto de $\lambda$.** Como $\nabla_\lambda\mathcal L=g$,

$$
\nabla_\lambda\mathcal L = 0
\qquad\Longleftrightarrow\qquad
g(x,m)=0
\qquad\text{(problema directo).}
$$ (eq:forward)

**(b) Estacionariedad respecto de $x$.** Como $\nabla_x\mathcal L=\nabla_xJ+\lambda^{T}D_xg$,

$$
\nabla_x\mathcal L = 0
\qquad\Longleftrightarrow\qquad
\big(D_xg\big)^{T}\lambda = -\big(\nabla_xJ\big)^{T}.
$$ (eq:adjoint)

Esta es la **ecuación adjunta**: un único sistema lineal de tamaño $n\times n$ que determina
$\lambda$. Notar que es la **traspuesta** del operador que aparece en el problema directo.

## 1.4 La derivada total y la cancelación del término de sensibilidad

Evaluamos ahora el Lagrangiano a lo largo de $x=x(m)$ y derivamos respecto de $m$. Como
$\lambda$ y $m$ son independientes pero $x$ depende de $m$, la regla de la cadena da

$$
\frac{d\hat J}{dm}
=\frac{d}{dm}\mathcal L\big(x(m),m,\lambda\big)
=\frac{\partial\mathcal L}{\partial m}
+\frac{\partial\mathcal L}{\partial x}\frac{dx}{dm}.
$$

Calculamos por separado cada derivada parcial, usando {eq}`eq:lagrangian`:

$$
\frac{\partial\mathcal L}{\partial m}
=\frac{\partial J}{\partial m}+\lambda^{T}\frac{\partial g}{\partial m}
\;\in\mathbb R^{1\times p},
\qquad
\frac{\partial\mathcal L}{\partial x}
=\frac{\partial J}{\partial x}+\lambda^{T}\frac{\partial g}{\partial x}
\;\in\mathbb R^{1\times n}.
$$ (eq:partials)

Pero la ecuación adjunta {eq}`eq:adjoint` dice exactamente que

$$
\frac{\partial J}{\partial x}+\lambda^{T}\frac{\partial g}{\partial x}=0_{1\times n}.
$$ (eq:cancel)

Por lo tanto el término que multiplica a la matriz de sensibilidad $\dfrac{dx}{dm}$ se anula
y la derivada total se reduce a

$$
\boxed{\;
\frac{d\hat J}{dm}
=\frac{\partial J}{\partial m}
+\lambda^{T}\frac{\partial g}{\partial m}
\;}
\qquad\text{con}\qquad
\big(D_xg\big)^{T}\lambda=-\big(\nabla_xJ\big)^{T}.
$$ (eq:gradient)

## 1.5 Costo del método

El gradiente {eq}`eq:gradient` se calcula con:

1. **una** resolución del problema directo $g(x,m)=0$;
2. **una** resolución del sistema adjunto $n\times n$, para obtener $\lambda$;
3. la evaluación de las derivadas parciales $\dfrac{\partial J}{\partial m}$ y
   $\dfrac{\partial g}{\partial m}$.

La matriz de sensibilidad $\dfrac{dx}{dm}$ **nunca se forma** y el costo es **independiente
de $p$**. Esta es la razón por la que el método adjunto es la columna vertebral de la
inversión y el diseño a gran escala.
