# 5. El punto de vista en dimensión infinita

Hasta aquí trabajamos **siempre en dimensión finita**: primero discretizamos la EDP y luego
aplicamos el método adjunto. En esta sección discutimos el camino alternativo —plantear el
adjunto **antes** de discretizar, en espacios de funciones— y comparamos ambos.

## 5.1 Teorema de Lagrange en espacios de Hilbert

Sea $V$ un espacio de Hilbert, $F:V\to\mathbb R$ y $G:V\to W$ ($W$ otro espacio de Hilbert)
aplicaciones diferenciables. Para el problema

$$
\min_{z\in V}\; F(z)
\qquad\text{sujeto a}\qquad
G(z)=0,
$$

bajo condiciones de regularidad adecuadas, existe un multiplicador $\mu\in W$ tal que

$$
\nabla F(z)+(D_zG)^{*}\mu=0,
$$ (eq:hilbert)

donde $(D_zG)^{*}:W\to V$ es el **operador adjunto** de $D_zG$ respecto de los productos
internos de $V$ y $W$. Este es el análogo en dimensión infinita de la ecuación adjunta
{eq}`eq:adjoint`.

## 5.2 Derivada de Gateaux

Para derivar funcionales en espacios de funciones usamos la **derivada de Gateaux**. Sea
$F:V\to\mathbb R$. La derivada de $F$ en $u$ en la dirección $h\in V$ es

$$
\delta F(u;h)=\left.\frac{d}{d\varepsilon}F(u+\varepsilon h)\right|_{\varepsilon=0}
=\lim_{\varepsilon\to0}\frac{F(u+\varepsilon h)-F(u)}{\varepsilon}.
$$

Si existe un elemento $\nabla F(u)\in V$ tal que

$$
\delta F(u;h)=\big\langle \nabla F(u),h\big\rangle
\qquad\text{para todo }h\in V,
$$ (eq:gateaux)

entonces $\nabla F(u)$ es el **gradiente** de $F$ respecto del producto interno
$\langle\cdot,\cdot\rangle$. Esto generaliza la identidad del cálculo en $\mathbb R^n$,

$$
\left.\frac{\partial}{\partial\varepsilon}f(x+\varepsilon d)\right|_{\varepsilon=0}
=\nabla f(x)\cdot d .
$$

**Ejemplo 5.1 (término de desajuste).** Con $F(u)=\displaystyle\int_0^1(u-u_d)^2\,dx$,

$$
\left.\frac{d}{d\varepsilon}F(u+\varepsilon h)\right|_{0}
=\int_0^1 2\,(u-u_d)\,h\,dx
=\big\langle 2(u-u_d),\,h\big\rangle
\quad\Longrightarrow\quad
\nabla_uF=2(u-u_d).
$$

**Ejemplo 5.2 (regularización).** Con $G(m)=\dfrac{\alpha}{2}\displaystyle\int_0^1m^2\,dx$,

$$
\left.\frac{d}{d\varepsilon}G(m+\varepsilon h)\right|_{0}
=\alpha\int_0^1 m\,h\,dx
=\big\langle\alpha m,\,h\big\rangle
\quad\Longrightarrow\quad
\nabla_mG=\alpha m .
$$

## 5.3 La ecuación adjunta por integración por partes

Repetimos el ejemplo de Poisson de la sección 3, pero **sin discretizar**. El Lagrangiano es

$$
\mathcal L(u,m,\lambda)=J(u,m)+\int_0^1\lambda\,(-u''-m)\,dx
=J(u,m)+\langle\lambda,-u''-m\rangle .
$$

Integramos por partes dos veces el término con $u''$:

$$
\int_0^1\lambda\,(-u'')\,dx
=-\big[\lambda\,u'\big]_0^1+\int_0^1\lambda' u'\,dx
=-\big[\lambda\,u'\big]_0^1+\big[\lambda' u\big]_0^1-\int_0^1\lambda'' u\,dx .
$$

Como $u(0)=u(1)=0$, el término $\big[\lambda' u\big]_0^1$ se anula. Entonces

$$
\mathcal L
=\int_0^1\Big[\big(u-u_d\big)^2-\lambda''u\Big]dx
+\frac{\alpha}{2}\int_0^1m^2\,dx
-\big[\lambda\,u'\big]_0^1
-\int_0^1\lambda\,m\,dx .
$$

La estacionariedad respecto de $u$ da, en el interior, $2(u-u_d)-\lambda''=0$; y para que
las variaciones de frontera no aporten (con $\delta u'$ libre) hay que exigir
$\lambda(0)=\lambda(1)=0$. Obtenemos así la **ecuación adjunta continua**, que es la misma
{\eq}`eq:poisson-adj-cont` identificada en la sección 3:

$$
-\lambda''=-2\big(u-u_d\big),
\qquad \lambda(0)=\lambda(1)=0 .
$$

El gradiente es $\nabla_mJ=\alpha m-\lambda$.

## 5.4 ¿Se llega al mismo resultado por los dos caminos?

Hay dos rutas para calcular el gradiente de un funcional gobernado por una EDP:

* **Discretizar y luego optimizar (DTO):** discretizar la EDP, formar el Lagrangiano
  *discreto* y derivar el adjunto (secciones 3 y 4).
* **Optimizar y luego discretizar (OTD):** derivar la ecuación adjunta *continua* (como en
  §5.3) y discretizarla para calcular el gradiente.

En general los dos gradientes **no coinciden exactamente**, aunque ambos son consistentes:
sólo difieren en $O(h^k)$ y convergen al mismo gradiente continuo cuando $h\to0$. Las
diferencias provienen de

1. **la cuadratura del funcional:** al discretizar $\int(u-u_d)^2$ aparece una **matriz de
   masa** $hI$ que hay que respetar al elegir el producto interno;
2. **las condiciones de borde:** el adjunto discreto impone las condiciones de frontera de la
   malla, mientras que la discretización del adjunto continuo puede tratarlas con otro orden;
3. **el operador discreto:** $A^{T}$ (o $A^{H}$) es la traspuesta **exacta** de la matriz
   discreta, y la discretización del adjunto continuo puede diferir en los bordes.

En el ejemplo de Poisson, **sí coinciden exactamente**. La razón es que elegimos el producto
interno discreto con la **matriz de masa** ($\langle a,b\rangle_h=h\,a^{T}b$): con esa
elección $A$ (simétrica) es el adjunto de sí misma y la ecuación discreta {eq}`eq:poisson-adj`
es idéntica a la discretización de la ecuación continua {eq}`eq:poisson-adj-cont`. Si en
cambio usáramos el producto interno euclídeo, el adjunto discreto sería
$A\lambda=-2h(u-u_d)$: el **mismo gradiente** salvo un reescalado del multiplicador, pero la
correspondencia con la ecuación continua queda sólo a menos de ese factor. En el ejemplo de
Helmholtz la coincidencia tampoco es exacta en los bordes (por la capa absorbente y el
tratamiento de la delta del foco), pero el adjunto discreto sigue siendo el gradiente
**exacto** del funcional discreto.

**Conclusión práctica.** En problemas reales (FWI, diseño inverso) se usa casi siempre
**DTO**: el gradiente discreto es la derivada **exacta** del funcional que efectivamente se
optimiza, de modo que las verificaciones contra diferencias finitas son exactas y no hay
sorpresas por condiciones de borde. La interpretación continua (OTD) es invaluable para
**entender** qué representa el adjunto —retropropagación, correlación cruzada, transporte de
sensibilidad— y para diseñar precondicionadores, pero para **calcular** el gradiente se
prefiere el adjunto discreto.
