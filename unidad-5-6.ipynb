# 6. Comentarios finales y ejercicios

## 6.1 ¿Por qué el adjunto es tan potente?

| Enfoque | Costo del gradiente |
|---|---|
| Diferencias finitas / sensibilidades | $O(p)$ resoluciones del problema directo |
| **Método adjunto** | $O(1)$: **un** directo + **un** adjunto por fuente/dato |

## 6.2 El patrón general

Válido en dimensión finita y en espacios de funciones:

1. Escribir el problema directo $g(x,m)=0$.
2. Elegir $\lambda$ resolviendo la **ecuación adjunta**
   $\big(D_xg\big)^{T}\lambda=-\big(\nabla_xJ\big)^{T}$ (o su versión continua,
   $(D_zG)^{*}\mu=-\nabla F$).
3. Evaluar $\dfrac{d\hat J}{dm}
   =\dfrac{\partial J}{\partial m}+\lambda^{T}\dfrac{\partial g}{\partial m}$.

## 6.3 En los ejemplos que hicimos

* **Control discreto:** el adjunto es una recurrencia hacia atrás en el tiempo (sección 2).
* **Poisson:** el adjunto discreto es la discretización de $-\lambda''=-2(u-u_d)$ con
  $\lambda(0)=\lambda(1)=0$ (sección 3).
* **Helmholtz:** el adjunto resuelve la misma ecuación de Helmholtz con una fuente en el foco
  y absorción conjugada; el gradiente es la **correlación cruzada** de los campos directo y
  retropropagado. La lente enfoca la onda plana en $\mathbf x_f=(14,7)$ (sección 4).

## 6.4 Ejercicios propuestos

1. Demuestre que si en la EDO de Poisson se impone una condición de Neumann, por ejemplo
   $u'(1)=0$, las condiciones de borde del adjunto **cambian**. ¿Cuáles son? Compárelas con
   las que da el adjunto discreto.
2. Implemente el **caso de datos** (FWI) para un arreglo de fuentes y receptores y compare el
   gradiente adjunto con el de diferencias finitas. Observe el parecido con el kernel de
   migración.
3. **Posición del foco:** repita el diseño de la lente para $\mathbf x_f=(12,7),(13,7),(14,7)$
   y compare la intensidad alcanzada. ¿Qué pasa si el foco está muy cerca de la lente?
4. **Compromiso focalización/suavidad:** barra el ancho de suavizado $s$ y grafique $J$ final
   contra la "rugosidad" de $\varepsilon$ (por ejemplo $\|\nabla\varepsilon\|^2$). ¿Dónde está
   el mejor compromiso?
